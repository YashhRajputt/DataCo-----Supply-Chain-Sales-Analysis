Converting string collumns to date time

```python
cols = ['shipping date (DateOrders)','order date (DateOrders)']
for col in cols:
    df[col] = pd.to_datetime(df[col],errors='coerce')
```

Count of Unique values in categorical columns

```python
cat_col = df.select_dtypes(include='object').columns
for col in cat_col:
    print(f'\n---{col:}---')
    print(df[col].value_counts(dropna=False).head(20))
```

Check categorical consistency

```python
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(col, df[col].astype(str).str.strip().ne(df[col].astype(str)).sum())
```

Count of Unique values in numerical columns

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

df.describe().T

```python
df.loc[
    df["Benefit per order"] == df["Benefit per order"].min(),
    ["Benefit per order", "Sales", "Order Item Discount",
     "Order Item Profit Ratio", "Order Id", "Product Name"]
]
```

```python
df.loc[
    df["Order Item Discount"] == df["Order Item Discount"].max(),
    ["Order Item Discount", "Order Item Discount Rate",
     "Order Item Product Price", "Product Price",
     "Sales", "Order Item Total", "Product Name"]
]
```

```python
df['Expected Discount'] = df['Order Item Discount Rate'] * df['Order Item Product Price']
df[["Order Item Discount", "Expected Discount"]].head(10)
```

```python
df["Discount Difference"] = (
    df["Order Item Discount"] -
    df["Expected Discount"]
)

df["Discount Difference"].describe()
```

```python
df["Profit Margin"] = (
    df["Order Profit Per Order"] / df["Sales"]
) * 100
```

```python
df['expected total'] = df['Order Item Product Price'] - df["Order Item Discount"]
df[['Order Item Total','expected total']].head()
```

```python
df['Delivery Delay'] = df['Days for shipping (real)'] - df['Days for shipment (scheduled)']
df["Delivery Delay"].value_counts().sort_index()
```

```python
pd.crosstab(
    df["Delivery Delay"] > 0,
    df["Late_delivery_risk"]
)
```

```python
df.drop(columns=['expected total','Discount Difference','Expected Discount'],inplace=True)
```

```python
df['Orders Month'] = df['order date (DateOrders)'].dt.month_name()
```

```python
df['Orders Year'] = df['order date (DateOrders)'].dt.to_period('Y')
```
